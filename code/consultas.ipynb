```python
import osmnx as ox
import matplotlib.pyplot as plt

# Descargar todos los barrios de Barcelona
tags = {"place": "neighbourhood"}

barrios = ox.features_from_place("Barcelona, España", tags)

print(barrios[["name", "geometry"]])
```

```python
fig, ax = plt.subplots(figsize=(10,10))

barrios.plot(
    ax=ax,
    edgecolor="black",
    facecolor="none",
    linewidth=0.8
)

plt.show()
```
